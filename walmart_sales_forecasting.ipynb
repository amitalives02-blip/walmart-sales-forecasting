# 4. Feature Engineering

## 4.1 Date Features

```python
merged_df["Year"] = merged_df["Date"].dt.year
merged_df["Month"] = merged_df["Date"].dt.month
merged_df["Week"] = merged_df["Date"].dt.isocalendar().week.astype(int)
merged_df["Quarter"] = merged_df["Date"].dt.quarter
merged_df["DayOfYear"] = merged_df["Date"].dt.dayofyear
```

```python
merged_df[["Date","Year","Month","Week","Quarter","DayOfYear"]].head()
```

## 4.2 Total MarkDown

```python
merged_df["Total_MarkDown"] = (
    merged_df["MarkDown1"]
    + merged_df["MarkDown2"]
    + merged_df["MarkDown3"]
    + merged_df["MarkDown4"]
    + merged_df["MarkDown5"]
)
```

```python
merged_df[[
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5",
    "Total_MarkDown"
]].head()
```

## 4.3 Lag Features

```python
# Sort before creating time-based features
merged_df = merged_df.sort_values(
    ["Store", "Dept", "Date"]
).reset_index(drop=True)

# Sales from the same store-department approximately 52 weeks earlier
merged_df["Lag52"] = (
    merged_df.groupby(["Store", "Dept"])["Weekly_Sales"]
    .shift(52)
)
```

```python
merged_df[[
    "Store",
    "Dept",
    "Date",
    "Weekly_Sales",
    "Lag52"
]].head(30)
```

## 4.4 Rolling Average

```python
merged_df["RollingMean4"] = (
    merged_df.groupby(["Store", "Dept"])["Weekly_Sales"]
    .transform(
        lambda series: series.shift(1).rolling(
            window=4,
            min_periods=1
        ).mean()
    )
)
```

```python
merged_df[[
    "Date",
    "Weekly_Sales",
    "RollingMean4"
]].head(10)
```

## 5.4 Train-Test Split

```python
train_df = merged_df[
    merged_df["Date"] < "2012-07-01"
].copy()

test_df = merged_df[
    merged_df["Date"] >= "2012-07-01"
].copy()
```

```python
train_df = train_df.drop(
    columns=["StoreDeptAvg"],
    errors="ignore"
)

test_df = test_df.drop(
    columns=["StoreDeptAvg"],
    errors="ignore"
)
```

```python
store_dept_average = (
    train_df
    .groupby(["Store", "Dept"])["Weekly_Sales"]
    .mean()
    .reset_index(name="StoreDeptAvg")
)

store_dept_average.head()
```

```python
train_df = train_df.merge(
    store_dept_average,
    on=["Store", "Dept"],
    how="left"
)

test_df = test_df.merge(
    store_dept_average,
    on=["Store", "Dept"],
    how="left"
)
```

```python
overall_training_average = train_df["Weekly_Sales"].mean()

train_df["StoreDeptAvg"] = (
    train_df["StoreDeptAvg"]
    .fillna(overall_training_average)
)

test_df["StoreDeptAvg"] = (
    test_df["StoreDeptAvg"]
    .fillna(overall_training_average)
)
```

```python
print("StoreDeptAvg in train_df:",
      "StoreDeptAvg" in train_df.columns)

print("StoreDeptAvg in test_df:",
      "StoreDeptAvg" in test_df.columns)

train_df[
    ["Store", "Dept", "Weekly_Sales", "StoreDeptAvg"]
].head()
```

```python
for dataframe in [train_df, test_df]:

    dataframe["Lag52"] = (
        dataframe["Lag52"]
        .fillna(dataframe["StoreDeptAvg"])
    )

    dataframe["RollingMean4"] = (
        dataframe["RollingMean4"]
        .fillna(dataframe["StoreDeptAvg"])
    )
```

```python
feature_columns = [
    "Store",
    "Dept",
    "IsHoliday",
    "Size",
    "Temperature",
    "Fuel_Price",
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5",
    "CPI",
    "Unemployment",
    "Year",
    "Month",
    "Week",
    "Quarter",
    "DayOfYear",
    "Total_MarkDown",
    "Lag52",
    "RollingMean4",
    "StoreDeptAvg",
    "Type_B",
    "Type_C"
]

x_train = train_df[feature_columns].copy()
y_train = train_df["Weekly_Sales"].copy()

x_test = test_df[feature_columns].copy()
y_test = test_df["Weekly_Sales"].copy()
```

```python
[column for column in train_df.columns
 if "StoreDeptAvg" in column]
```

```python
print("Training period:", train_df["Date"].min(), "to", train_df["Date"].max())
print("Testing period :", test_df["Date"].min(), "to", test_df["Date"].max())

print("\nTraining shape:", x_train.shape)
print("Testing shape :", x_test.shape)

required_columns = ["Lag52", "RollingMean4", "StoreDeptAvg"]
print("\nMissing values in training features:")
print(train_df[required_columns].isnull().sum())

print("\nMissing values in testing features:")
print(test_df[required_columns].isnull().sum())
```

## 5.5 Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = x_train.copy()
x_test_scaled = x_test.copy()

x_train_scaled[numerical_features] = scaler.fit_transform(
    x_train[numerical_features]
)

x_test_scaled[numerical_features] = scaler.transform(
    x_test[numerical_features]
)
```

```python
sales_history = train_df[
    ["Store", "Dept", "Date", "Weekly_Sales"]
].copy()

sales_history["Date"] = pd.to_datetime(sales_history["Date"])

store_information = (
    train_df[
        ["Store", "Size", "Type_B", "Type_C"]
    ]
    .drop_duplicates(subset=["Store"])
    .sort_values("Store")
    .reset_index(drop=True)
)

store_dept_lookup = (
    train_df[
        ["Store", "Dept", "StoreDeptAvg"]
    ]
    .drop_duplicates(subset=["Store", "Dept"])
    .reset_index(drop=True)
)

default_values = {
    "Temperature": float(train_df["Temperature"].median()),
    "Fuel_Price": float(train_df["Fuel_Price"].median()),
    "CPI": float(train_df["CPI"].median()),
    "Unemployment": float(train_df["Unemployment"].median()),
    "MarkDown1": 0.0,
    "MarkDown2": 0.0,
    "MarkDown3": 0.0,
    "MarkDown4": 0.0,
    "MarkDown5": 0.0
}

print("Sales-history rows:", len(sales_history))
print("Stores available:", len(store_information))
print("Store-department combinations:", len(store_dept_lookup))
```

## 8.2 Save the Trained Model Bundle


```python
import joblib

model_bundle = {
    "model": xgb_model,
    "feature_columns": feature_columns,
    "sales_history": sales_history,
    "store_information": store_information,
    "store_dept_lookup": store_dept_lookup,
    "default_values": default_values,
    "overall_training_average": float(overall_training_average),
    "model_metrics": {
        "MAE": float(xgb_mae),
        "RMSE": float(xgb_rmse),
        "R2": float(xgb_r2)
    }
}

joblib.dump(
    model_bundle,
    "walmart_xgb_bundle.joblib",
    compress=3
)

print("Saved: walmart_xgb_bundle.joblib")
```
